```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from qplay import nuclear

U_MEV = 931.494          # energy equivalent of 1 u, in MeV
U_KG = 1.66054e-27       # 1 u in kg
MEV_J = 1.602e-13        # 1 MeV in J

# Measured masses, in u
m_p, m_n = 1.007276467, 1.008664916      # proton, neutron
m_He4_nucleus = 4.001506179
m_H1_atom, m_Fe56_atom = 1.00782503, 55.93493633
```

## 1. Mass defect and binding energy

A nucleus is **lighter** than the protons and neutrons it is made from. The difference is the **mass defect**,

$$\Delta m = Zm_p + (A - Z)m_n - m_{\text{nucleus}}.$$

To pull the nucleus apart into separate nucleons you must supply energy, the **binding energy**, and by Einstein's $E = mc^2$ that energy is exactly the missing mass:

$$E_B = \Delta m\,c^2 .$$

Nuclear masses are given in atomic mass units, u, and $1\ \mathrm{u}$ is equivalent to $931.5\ \mathrm{MeV}$. So a mass defect in u, times 931.5, is the binding energy in MeV.

Tables usually give the masses of whole **atoms**, electrons included. Use the mass of a hydrogen-1 atom in place of $m_p$: then the $Z$ electrons appear on both sides and cancel.

```python
defect_He = 2 * m_p + 2 * m_n - m_He4_nucleus
E_He = defect_He * U_MEV
print(f"helium-4: mass defect {defect_He:.6f} u, binding energy {E_He:.2f} MeV, {E_He / 4:.3f} MeV per nucleon")

defect_Fe = 26 * m_H1_atom + 30 * m_n - m_Fe56_atom      # atomic masses: the 26 electrons cancel
E_Fe = defect_Fe * U_MEV
print(f"iron-56:  mass defect {defect_Fe:.6f} u, binding energy {E_Fe:.1f} MeV, {E_Fe / 56:.3f} MeV per nucleon")
print(f"the defect is {100 * defect_Fe / m_Fe56_atom:.2f}% of the mass of an iron-56 atom")
```

The total binding energy grows with the number of nucleons, so to compare nuclei we use the **binding energy per nucleon**, $E_B/A$. It measures how tightly, on average, each nucleon is held: the higher it is, the more stable the nucleus.

## 2. The binding-energy curve

The cell below plots measured values of $E_B/A$ for nuclei from hydrogen-2 to uranium-238 (AME2020 atomic mass evaluation, rounded), together with the liquid-drop model from section 4.

**Predict (P1).** Does the binding energy per nucleon keep increasing as nuclei get bigger, level off, or rise and then fall? Sketch your prediction before you run the next cell.

```python
measured = {  # nucleus: (A, binding energy per nucleon in MeV), measured
    "H-2": (2, 1.112), "He-4": (4, 7.074), "Li-7": (7, 5.606), "C-12": (12, 7.680), "O-16": (16, 7.976),
    "Fe-56": (56, 8.790), "Ni-62": (62, 8.795), "Kr-84": (84, 8.717), "Sn-120": (120, 8.505),
    "Ba-138": (138, 8.393), "Pb-208": (208, 7.868), "U-235": (235, 7.591), "U-238": (238, 7.570),
}
A_line = np.arange(10, 251)
Z_line = [int(round(A / (1.98 + 0.0155 * A ** (2 / 3)))) for A in A_line]     # near the line of stable nuclei
model = [nuclear.binding_energy_per_nucleon(int(A), Z) for A, Z in zip(A_line, Z_line)]

plt.figure(figsize=(8, 4.2))
plt.plot(A_line, model, color="0.6", label="liquid-drop model")
for name, (A, b) in measured.items():
    plt.plot(A, b, "o", color="C0")
    offset = {"Li-7": (4, -12), "U-238": (4, -12), "C-12": (8, -4), "He-4": (4, -12), "O-16": (-16, 8), "Fe-56": (-30, 6), "Ni-62": (2, -13)}
    plt.annotate(name, (A, b), textcoords="offset points", xytext=offset.get(name, (4, 5)), fontsize=8)
plt.plot([], [], "o", color="C0", label="measured")
plt.axvspan(50, 66, color="C2", alpha=0.12, label="most tightly bound (iron, nickel)")
plt.xlabel("nucleon number A"); plt.ylabel("binding energy per nucleon / MeV"); plt.ylim(0, 9.6)
plt.legend(loc="lower right", fontsize=8); plt.title("Binding energy per nucleon");

peak = max(measured, key=lambda name: measured[name][1])
print(f"most tightly bound in the table: {peak} at {measured[peak][1]} MeV per nucleon")
```

**Explain (P1).** Every nucleon attracts only its nearest neighbours through the strong force, but every proton repels every other proton. Use this to explain the rise for small nuclei and the slow fall for large ones.

## 3. Fission and fusion

Any change that moves nuclei **towards the peak** of the curve releases energy, because the products are more tightly bound than what you started with. The energy released is

$$Q = (\text{total binding energy of the products}) - (\text{total binding energy of the reactants}),$$

equal to $\Delta m\,c^2$ for the mass that disappears.

- **Fission** splits a heavy nucleus such as uranium-235 into two medium nuclei, moving down the right-hand side of the curve towards the peak.
- **Fusion** joins light nuclei such as hydrogen-2 (deuterium) and hydrogen-3 (tritium), climbing the steep left-hand side.

Free neutrons have no binding energy, so they contribute nothing to the totals.

**Predict (P2).** Kilogram for kilogram of fuel, which releases more energy: the fission of uranium-235 or the fusion of deuterium and tritium?

```python
# Fusion: H-2 + H-3 -> He-4 + n, using measured total binding energies in MeV
Q_fusion = 28.2957 - (2.2246 + 8.4818)
# Fission: n + U-235 -> Ba-141 + Kr-92 + 3n, using binding energies per nucleon
Q_fission = 141 * 8.33 + 92 * 8.51 - 235 * 7.59

for name, Q, mass_in_u in [("D-T fusion", Q_fusion, 5), ("U-235 fission", Q_fission, 236)]:
    per_kg = Q * MEV_J / (mass_in_u * U_KG)
    print(f"{name:>13}: {Q:6.1f} MeV per reaction, {Q / mass_in_u:.2f} MeV per u of fuel, {per_kg:.2e} J/kg")
print(f"fusion releases {(Q_fusion / 5) / (Q_fission / 236):.1f} times as much energy per kilogram")

labels = ["burning coal", "U-235 fission", "D-T fusion"]
energies = [3e7, Q_fission * MEV_J / (236 * U_KG), Q_fusion * MEV_J / (5 * U_KG)]
plt.figure(figsize=(6, 3))
plt.barh(labels, energies, color=["0.6", "C0", "C1"])
plt.xscale("log"); plt.xlabel("energy released per kilogram of fuel / J (log scale)")
plt.title("Chemical against nuclear energy");
```

**Explain (P2).** Fission releases about ten times more energy per reaction. Why does fusion still win per kilogram? Look at how steep each side of the binding-energy curve is.

Both are millions of times more than chemistry: burning a kilogram of coal releases about $3\times10^{7}$ J, the grey bar.

## 4. The liquid-drop model

The curve can be explained by treating the nucleus as a drop of incompressible nuclear fluid. The **semi-empirical mass formula** used by `nuclear.binding_energy_per_nucleon` adds up five terms, in MeV:

$$E_B = \underbrace{15.75A}_{\text{volume}} - \underbrace{17.8A^{2/3}}_{\text{surface}} - \underbrace{0.711\frac{Z(Z-1)}{A^{1/3}}}_{\text{Coulomb}} - \underbrace{23.7\frac{(A-2Z)^2}{A}}_{\text{asymmetry}} + \underbrace{\delta}_{\text{pairing}} .$$

- **Volume:** each nucleon is bound to its neighbours by the short-range strong force, so binding grows in proportion to $A$.
- **Surface:** nucleons at the surface have fewer neighbours. The surface area grows as $A^{2/3}$, so this matters most for small nuclei.
- **Coulomb:** every proton repels every other, so this grows roughly as $Z^2$ and dominates in heavy nuclei.
- **Asymmetry:** nuclei with unequal numbers of protons and neutrons are less tightly bound.
- **Pairing:** $\delta = +12/\sqrt{A}$ MeV if $Z$ and $N$ are both even, $-12/\sqrt{A}$ if both are odd, and zero if $A$ is odd.

The coefficients are fitted to measured masses. The model is good for medium and heavy nuclei but poor for very light ones: it gives helium-4 only 5.81 MeV per nucleon, against the measured 7.07.

**Predict (P3).** Three nuclei have the same nucleon number, 56: manganese-56 ($Z = 25$), iron-56 ($Z = 26$) and cobalt-56 ($Z = 27$). Which does the model bind most tightly, and why?

```python
A = 56
Z = np.arange(int(0.35 * A), int(0.5 * A) + 7)          # wide enough to contain the most bound Z for any A
per_nucleon = [nuclear.binding_energy_per_nucleon(A, int(z)) for z in Z]
for z in (25, 26, 27) if A == 56 else ():
    print(f"Z = {z} (N = {A - z}): {nuclear.binding_energy_per_nucleon(A, z):.3f} MeV per nucleon")
best = int(Z[np.argmax(per_nucleon)])
print(f"most tightly bound for A = {A}: Z = {best} (N = {A - best})")

plt.figure(figsize=(7, 3.6))
plt.plot(Z, per_nucleon, "o-")
plt.annotate(f"Z = {best}", (best, max(per_nucleon)), textcoords="offset points", xytext=(-22, 7))
plt.ylim(top=9.0)
plt.xlabel("proton number Z"); plt.ylabel("binding energy per nucleon / MeV")
plt.title(f"Nuclei with A = {A}: the valley of stability");
```

**Explain (P3).** Which terms change when $Z$ changes but $A$ stays at 56? Why does the zigzag between neighbouring points appear?

*Try it:* change `A` to 100 or 208. Is the most tightly bound $Z$ the stable nucleus you would find in a data book? Is it half of $A$?

**Common mistakes**

- Thinking binding energy is energy the nucleus *has*. It is the energy you must **supply** to pull it apart; a nucleus with more binding energy has **less** mass.
- Comparing total binding energies instead of binding energy **per nucleon** when judging stability.
- Mixing up nuclear and atomic masses: with atomic masses, use the hydrogen-1 atom for the protons so that the electrons cancel.
- Forgetting to multiply by the number of nucleons before subtracting total binding energies in a reaction.

### Predict and explain: answers

**P1.** It rises steeply, peaks near iron and nickel ($A \approx 56$ to 62, about 8.8 MeV per nucleon), then falls slowly. Small nuclei have a large fraction of their nucleons at the surface with fewer neighbours, so adding nucleons increases the binding per nucleon. In large nuclei the strong force still reaches only neighbours, but the electrostatic repulsion acts between every pair of protons, so it grows faster than the binding and $E_B/A$ falls.

**P2.** D-T fusion, by about five times per kilogram, even though one fission releases about ten times more energy than one fusion. A fission event needs 236 u of fuel; a fusion event only 5 u. The left-hand side of the curve is much steeper, so each light nucleon gains far more binding energy when light nuclei fuse.

**P3.** Iron-56. With $A$ fixed, the volume and surface terms are the same for all three; only the Coulomb, asymmetry and pairing terms change. More protons means more repulsion, fewer protons means a bigger neutron–proton imbalance, and iron-56 strikes the best balance. It is also even-even (26 protons, 30 neutrons), so it gains the pairing bonus, while manganese-56 and cobalt-56 are odd-odd and lose it. The pairing term makes the zigzag between neighbouring points.
