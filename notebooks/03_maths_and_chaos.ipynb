# 03 · Maths and chaos

Numerical methods, Fourier series and a tour of chaos.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from qplay import numerics, waves, chaos
```

## Newton's method

Solve $x^2 - 2 = 0$ by iterating $x_{n+1} = x_n - f(x_n)/f'(x_n)$, which converges quadratically.

```python
x = 1.0
for i in range(6):
    x = x - (x * x - 2) / (2 * x)
    print(i, x, abs(x - np.sqrt(2)))
```

## Fourier series of a square wave

$$\operatorname{sgn}(\sin x) = \frac{4}{\pi}\sum_{k\ \mathrm{odd}}\frac{\sin kx}{k}.$$

Note the overshoot near the jumps, the Gibbs phenomenon, which never goes away however many terms you add.

```python
period = 2 * np.pi
a0, a, b = waves.fourier_coefficients(lambda s: np.sign(np.sin(s)), period, 50, points=[np.pi])
x = np.linspace(0, 2 * np.pi, 1000)
for n in (1, 5, 49):
    plt.plot(x, waves.fourier_sum(a0, a[:n], b[:n], x, period), label=f"{n} terms")
plt.legend();
```

## The road to chaos

For the logistic map $x_{n+1}=r\,x_n(1-x_n)$ the Lyapunov exponent $\lambda$ is negative for stable orbits and positive for chaos.

```python
r, xs = chaos.bifurcation(np.linspace(2.8, 4.0, 1000))
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7, 6), sharex=True)
ax1.scatter(r, xs, s=0.1, c="k", alpha=0.3, linewidths=0)
rs = np.linspace(2.8, 4.0, 300)
ax2.plot(rs, [chaos.lyapunov_logistic(v, n=2000) for v in rs], lw=0.8)
ax2.axhline(0, color="r", lw=0.5); ax2.set_xlabel("r"); ax2.set_ylabel(r"$\lambda$");
```

## Fractals

The Mandelbrot set is the set of $c$ for which $z \to z^2 + c$ stays bounded when started from $z=0$.

```python
plt.figure(figsize=(8, 5))
plt.imshow(chaos.mandelbrot(width=800, height=520, max_iter=150), extent=[-2.2, 0.8, -1.2, 1.2], cmap="magma", origin="lower")
plt.grid(False);
```

```python
plt.figure(figsize=(7, 5))
plt.imshow(chaos.julia(-0.8 + 0.156j, max_iter=150), extent=[-1.6, 1.6, -1.2, 1.2], cmap="twilight_shifted", origin="lower")
plt.grid(False);
```
